```python
# 1. Crear el catálogo y el esquema si no existen
spark.sql("CREATE CATALOG IF NOT EXISTS big_data_team1")
spark.sql("CREATE SCHEMA IF NOT EXISTS big_data_team1.mathias")

# 2. Definir la ruta pública de tu archivo Silver en S3
ruta = "s3://amankay-datalake-calidad-2026/silver/ingemmet/2026-10-21/ingemmet_silver_2026-10-21.csv"

# 3. Leer el CSV especificando el delimitador punto y coma (;)
df = (
    spark.read
    .option("header", "true")
    .option("sep", ";")
    .option("inferSchema", "true")
    .csv(ruta)
)

# 4. Guardar y registrar la tabla en Unity Catalog
df.write.mode("overwrite").saveAsTable("big_data_team1.mathias.ingemmet_silver")

# 5. Mostrar los primeros 10 registros
display(df.limit(10))
```
